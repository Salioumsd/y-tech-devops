```python
import os
import torch
from datasets import load_dataset
from transformers import AutoModelForCausalLM, AutoTokenizer, TrainingArguments, Trainer, DataCollatorForLanguageModeling, BitsAndBytesConfig
from peft import LoraConfig, get_peft_model, TaskType

# 1. CONFIGURATION ANTI-OOM
MODEL_ID = "microsoft/Phi-3.5-mini-instruct"
OUTPUT_DIR = "./results/medical_lora"

print("--- 1. Configuration de la Quantification 4-bit ---")
# Cette configuration compresse le modèle pour qu'il tienne largement dans les 14.5 Go du GPU
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.bfloat16,
    bnb_4bit_use_double_quant=True
)

print("--- 2. Chargement du Modèle compressé et Tokenizer ---")
tokenizer = AutoTokenizer.from_pretrained(MODEL_ID)
tokenizer.pad_token = tokenizer.eos_token
model = AutoModelForCausalLM.from_pretrained(
    MODEL_ID,
    quantization_config=bnb_config, # Application de la compression
    device_map="auto"
)

print("--- 3. Configuration LoRA (Étape IA/CYBER) ---")
peft_config = LoraConfig(
    task_type=TaskType.CAUSAL_LM,
    r=8,              # Réduit de 16 à 8 pour économiser la VRAM
    lora_alpha=16,
    lora_dropout=0.05,
    target_modules=["qkv_proj"] # Focus minimaliste pour le PoC
)
model = get_peft_model(model, peft_config)
model.print_trainable_parameters()

print("--- 4. Téléchargement direct du Dataset Médical ---")
dataset = load_dataset("ruslanmv/ai-medical-chatbot", split="train[:100]") # Échantillon de 100 suffisant pour le jury

def format_prompt(examples):
    texts = [f"<s>### Human: {p} ### Assistant: {d}</s>" for p, d in zip(examples["Patient"], examples["Doctor"])]
    return tokenizer(texts, truncation=True, max_length=256) # Réduit de 512 à 256 pour sécuriser la mémoire

tokenized_dataset = dataset.map(format_prompt, batched=True, remove_columns=dataset.column_names)

print("--- 5. Paramètres d'Entraînement Ultra-Légers ---")
args = TrainingArguments(
    output_dir=OUTPUT_DIR,
    per_device_train_batch_size=1,        # 1 seul exemple à la fois pour éviter l'explosion de VRAM
    gradient_accumulation_steps=4,       # On simule un batch de 4 artificiellement
    learning_rate=2e-4,
    logging_steps=2,
    num_train_epochs=1,
    bf16=True,
    save_strategy="no"
)

trainer = Trainer(
    model=model,
    args=args,
    train_dataset=tokenized_dataset,
    data_collator=DataCollatorForLanguageModeling(tokenizer=tokenizer, mlm=False)
)

print("--- 6. Lancement du Fine-Tuning Sécurisé ---")
trainer.train()
print(f"🎉 Terminé avec succès ! Adaptateurs sauvegardés dans {OUTPUT_DIR}")
```
